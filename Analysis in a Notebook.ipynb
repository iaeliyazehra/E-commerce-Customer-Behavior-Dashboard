### **Step 4: Outliers**

Outliers are unusually high or low values that may affect data analysis.

### Box Plot of Numeric Columns

A box plot helps identify potential outliers in the numerical columns.

```python
numeric_values = df.select_dtypes(include="number")
plt.figure(figsize=(7,15))
sns.boxplot(data=numeric_values, color="#2563eb")
plt.xticks(rotation=45, fontsize=10)
plt.title("Box Plot of Numeric Columns")
plt.show()
```

### Histogram of Numeric Columns

Histograms show the distribution of values in each numerical column and help identify the spread and skewness of the data.

```python
for column in numeric_values :
  plt.figure(figsize=(8,4))
  plt.hist(numeric_values[column], bins=20, edgecolor="black")
  plt.title(f"Distribution of {column}")
  plt.xlabel(column)
  plt.ylabel("Frequency")
  plt.show
```

### Outlier Detection Using IQR

The IQR method is used to identify potential outliers in the dataset. We will apply it to two numerical columns.

```python
columns = ["Yearly Amount Spent", "Length of Membership" ]

for column in columns:
  Q1 = df[column].quantile(.25)
  Q3 = df[column].quantile(.75)

  IQR = Q3 - Q1

  lower_limit = Q1 - 1.5 * IQR
  upper_limit = Q3 + 1.5 * IQR

  Outliers = df[(df[column] < lower_limit) | (df[column] > upper_limit)]

  print(column)
  print("Number of Outliers:",len(Outliers))
  print()
```

The outliers were kept because they may be real customer values. Removing them could result in losing useful data.

**Q3. Which customer activity is most strongly associated with yearly spending?**

**Q8. Which factor has the strongest relationship with yearly spending?**



```python
correlation = df[["Avg. Session Length", "Time on App", "Time on Website", "Length of Membership"]].corrwith(df["Yearly Amount Spent"])

plt.figure(figsize=(6,4))
correlation.plot(kind="bar")

plt.title("Factors Related to Yearly Spending")
plt.xlabel("Factors")
plt.ylabel("Correlation")
plt.show()
```
